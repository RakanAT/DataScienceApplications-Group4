```python
import requests
import os

# Define books and their corresponding Project Gutenberg IDs
# IDs in the URL of the book on Gutenberg.org
books_to_download = {
    #"TheWonderfulWizardOfOz": "55",
    "TheGodsOfPegana": "10827",
    "TheEnchantedCastle": "836",
    "TheKingOfElflandsDaughter": "10023",
    #"PrideAndPrejudice": "1342",
    "JaneEyre": "1260",
    "ARoomWithAView": "2641",
    "TheBlueCastle": "46904",
    #"OnTheTrailOfTheSpacePirates": "19409",
    "TheTimeMachine": "35",
    "APrincessOfMars": "62",
    "TwentyThousandLeagues": "164",
    #"Frankenstein": "84",
    "Dracula": "345",
    "JekyllAndHyde": "42",
    "KingInYellow": "8492",
    #"AutobiographyOfBenjaminFranklin": "20203",
    "LifeOnTheMississippi": "245",
    "NarrativeOfFrederickDouglass": "23",
    "ConfessionsOfStAugustine": "3296"
}

def download_gutenberg_book(filename, book_id):
    url = f"https://www.gutenberg.org/files/{book_id}/{book_id}-0.txt"
    # Fallback to older format if the -0.txt doesn't exist
    try:
        response = requests.get(url)
        if response.status_code != 200:
            url = f"https://www.gutenberg.org/ebooks/{book_id}.txt.utf-8"
            response = requests.get(url)
        
        with open(f"{filename}.txt", "w", encoding="utf-8") as f:
            f.write(response.text)
        print(f"Successfully downloaded: {filename}")
    except Exception as e:
        print(f"Failed to download {filename}: {e}")

# Run the download
if __name__ == "__main__":
    for name, book_id in books_to_download.items():
        download_gutenberg_book(name, book_id)
```
